```python
male_dead=((data['Sex'] == 'male') & (data['Survived'] ==0 )).sum()
print('Male deaths : ',male_dead)
female_dead=((data['Sex'] == 'female') & (data['Survived'] ==0 )).sum()
print('Female deaths : ',female_dead)
male_survived=((data['Sex'] == 'male') & (data['Survived'] == 1)).sum()
print('Male Survived : ',male_dead)
female_survived=((data['Sex'] == 'female') & (data['Survived'] ==1)).sum()
print('Female Survived : ',female_dead)
```

```python
labels = ['Male', 'Female']
deaths = [male_dead, female_dead]
survivals = [male_survived, female_survived]

plt.figure(figsize=(8, 6))
plt.bar(labels, deaths, color='salmon', label='Died')
plt.bar(labels, survivals, bottom=deaths, color='skyblue', label='Survived')  

plt.title('Survival vs Death Count by Gender')
plt.ylabel('Number of Passengers')
plt.legend()
plt.tight_layout()
plt.show()
```
